```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler


columns = (
    ["engine_id", "cycle", "setting_1", "setting_2", "setting_3"] +
    [f"sensor_{i}" for i in range(1, 22)]
)

train = pd.read_csv(
    "../dataset/CMAPSSData/train_FD001.txt",
    sep=r"\s+",
    header=None,
    names=columns
)

print("Missing values:", train.isnull().sum().sum())
print("Duplicate rows:", train.duplicated().sum())

train = train.fillna(train.median(numeric_only=True))
train = train.drop_duplicates()

train["stage"] = pd.cut(
    train["cycle"],
    bins=3,
    labels=["Early", "Middle", "Late"]
)

train = pd.get_dummies(train, columns=["stage"], dtype=int)

sensors = [f"sensor_{i}" for i in range(1, 22)]

train[sensors] = MinMaxScaler().fit_transform(train[sensors])
print("\nNormalized data:")
print(train[sensors].head())

train[sensors] = StandardScaler().fit_transform(train[sensors])
print("\nStandardized data:")
print(train[sensors].head())

print("Experiment 2 Completed Successfully")
```
