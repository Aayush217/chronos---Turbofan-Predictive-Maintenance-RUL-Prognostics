```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


cols = ["engine_id","cycle","setting_1","setting_2","setting_3"] + [f"sensor_{i}" for i in range(1,22)]

df = pd.read_csv("../dataset/CMAPSSData/train_FD001.txt",
                 sep=r"\s+", header=None, names=cols)

print(df.describe())

sns.histplot(df["sensor_2"])
plt.title("Sensor 2 Distribution")
plt.show()

sns.scatterplot(x=df["sensor_2"], y=df["sensor_3"], s=10)
plt.title("Sensor 2 vs Sensor 3")
plt.show()

sns.boxplot(x=df["sensor_2"])
plt.title("Sensor 2 Box Plot")
plt.show()

sns.heatmap(df.iloc[:, 5:12].corr(), annot=True)
plt.title("Sensor Correlation")
plt.show()

print("Experiment 3 Completed Successfully")
```
