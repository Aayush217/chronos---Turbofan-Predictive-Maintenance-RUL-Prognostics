```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


columns = (
    ["engine_id", "cycle", "setting_1", "setting_2", "setting_3"] +
    [f"sensor_{i}" for i in range(1, 22)]
)

train = pd.read_csv(
    "../dataset/CMAPSSData/train_FD001.txt",
    sep=r"\s+",
    header=None,
    names=columns
)

test = pd.read_csv(
    "../dataset/CMAPSSData/test_FD001.txt",
    sep=r"\s+",
    header=None,
    names=columns
)

print("Training Data:")
print(train.head())

print("\nTraining shape:", train.shape)
print("Testing shape:", test.shape)

print("\nMean of Sensor 1:", np.mean(train["sensor_1"]))

engine_1 = train[train["engine_id"] == 1]

plt.plot(engine_1["cycle"], engine_1["sensor_2"])
plt.xlabel("Engine Cycle")
plt.ylabel("Sensor 2")
plt.title("Engine 1 - Sensor 2")
plt.show()

print("Experiment 1 Completed Successfully")
```
