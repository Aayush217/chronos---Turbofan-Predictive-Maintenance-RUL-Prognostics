```python
import pandas as pd
from sklearn.linear_model import LinearRegression


cols = ["engine_id","cycle","setting_1","setting_2","setting_3"] + [f"sensor_{i}" for i in range(1,22)]

df = pd.read_csv("../dataset/CMAPSSData/train_FD001.txt",
                 sep=r"\s+", header=None, names=cols)

df["RUL"] = df.groupby("engine_id")["cycle"].transform("max") - df["cycle"]

# Simple Linear Regression
model1 = LinearRegression()
model1.fit(df[["cycle"]], df["RUL"])

print("\nSimple Regression Coefficient:", model1.coef_[0])
print("Predicted RUL:", model1.predict(pd.DataFrame({"cycle": [50]}))[0])

# Multiple Linear Regression
features = ["cycle", "sensor_2", "sensor_3"]
model2 = LinearRegression()
model2.fit(df[features], df["RUL"])

print("\nMultiple Regression Coefficients:")
print(pd.Series(model2.coef_, index=features))

print("Experiment 4 Completed Successfully")
```
